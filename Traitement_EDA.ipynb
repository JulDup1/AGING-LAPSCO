```python
import os
import pandas as pd
import shutil
import neurokit2 as nk
import matplotlib.pyplot as plt
```

## Filtre 2400 valeurs

```python
# 2400 valeurs correspond à 10 minutes à 4 HZ

dossiers = [r"C:\Users\judupont\Desktop\EDA_v2_tronque_blocs", r"C:\Users\judupont\Desktop\EDA_v4_tronque_blocs"]

for dossier in dossiers:
    
    print(f"\n===== Vérification du dossier : {dossier} =====\n")
    
    nb_total = 0
    nb_valides = 0
    fichiers_insuffisants = []

    for fichier in os.listdir(dossier):
        if fichier.endswith(".csv"):
            
            chemin = os.path.join(dossier, fichier)
            df = pd.read_csv(chemin)
            
            n_lignes = len(df)
            nb_total += 1
            
            if n_lignes >= 2400:
                nb_valides += 1
            else:
                fichiers_insuffisants.append((fichier, n_lignes))

    print(f"Nombre total de fichiers : {nb_total}")
    print(f"Fichiers ≥ 2400 lignes : {nb_valides}")
    print(f"Fichiers rejetés : {len(fichiers_insuffisants)}\n")

    if fichiers_insuffisants:
        print("⚠️ Fichiers avec longueur insuffisante :\n")
        for f, n in fichiers_insuffisants:
            print(f"{f} → {n} lignes")
```

```python
dossier_source = r"C:\Users\judupont\Desktop\EDA_v2_tronque_blocs"
dossier_sortie = r"C:\Users\judupont\Desktop\EDA_v2_blocs_valides"

os.makedirs(dossier_sortie, exist_ok=True)

candidats_rejetes = set()

# -------------------------
# Étape 1 : détecter les blocs trop courts
# -------------------------

for fichier in os.listdir(dossier_source):

    if fichier.endswith(".csv"):
        
        chemin = os.path.join(dossier_source, fichier)
        df = pd.read_csv(chemin)
        
        n_lignes = len(df)

        if n_lignes < 2400:
            candidat = fichier[:7]  # numéro inclusion
            candidats_rejetes.add(candidat)

# -------------------------
# Étape 2 : copier seulement les candidats valides
# -------------------------

for fichier in os.listdir(dossier_source):

    if fichier.endswith(".csv"):
        
        candidat = fichier[:7]

        if candidat not in candidats_rejetes:
            
            src = os.path.join(dossier_source, fichier)
            dst = os.path.join(dossier_sortie, fichier)
            
            shutil.copy(src, dst)

# -------------------------
# Étape 3 : afficher les rejetés
# -------------------------

print("\n⚠️ Candidats rejetés (bloc < 2400 lignes) :\n")

for c in sorted(candidats_rejetes):
    print(c)

print(f"\nNombre total de candidats rejetés : {len(candidats_rejetes)}")
```

## Calculs min, max, mean, std

```python
import os
import pandas as pd

dossier = r"C:\Users\judupont\Desktop\EDA_v2_blocs_valides"
chemin_sortie = r"C:\Users\judupont\Desktop\EDA_v2_stats.csv"

data = {}

for fichier in os.listdir(dossier):

    if fichier.endswith(".csv"):

        chemin = os.path.join(dossier, fichier)
        df = pd.read_csv(chemin)

        signal = df.iloc[:,0]

        candidat = fichier[:7]

        # récupération du numéro de bloc
        if "bloc1" in fichier:
            bloc = "bloc1"
        elif "bloc2" in fichier:
            bloc = "bloc2"
        elif "bloc3" in fichier:
            bloc = "bloc3"
        else:
            continue

        stats = {
            f"{bloc}_min": signal.min(),
            f"{bloc}_max": signal.max(),
            f"{bloc}_mean": signal.mean(),
            f"{bloc}_std": signal.std()
        }

        if candidat not in data:
            data[candidat] = {}

        data[candidat].update(stats)

# création du dataframe 
df_eda_v2 = pd.DataFrame.from_dict(data, orient="index")

df_eda_v2.reset_index(inplace=True)
df_eda_v2.rename(columns={"index": "Numero_inclusion"}, inplace=True)
df_eda_v2 = df_eda_v2.round(2)

df_eda_v2
df_eda_v2.to_csv(chemin_sortie, index=False, encoding="utf-8-sig")
```

## Calculs SCL et SCR

```python
dossier = r"C:\Users\judupont\Desktop\EDA_v2_blocs_valides"
dossier_sortie = r"C:\Users\judupont\Desktop\EDA_output_V2"

os.makedirs(dossier_sortie, exist_ok=True)

fichiers = [f for f in os.listdir(dossier) if f.endswith(".csv")]

# ============================
# Regrouper par candidat
# ============================
candidats = {}

for fichier in fichiers:
    candidat = fichier[:7]
    candidats.setdefault(candidat, []).append(fichier)

# ============================
# Dictionnaire final
# ============================
data_dict = {}

nb_candidats_ok = 0

# ============================
# Traitement
# ============================
for candidat, fichiers_candidat in candidats.items():

    if len(fichiers_candidat) != 3:
        print(f"❌ {candidat} rejeté")
        continue

    print(f"✅ {candidat} OK")

    dossier_candidat = os.path.join(dossier_sortie, candidat)
    os.makedirs(dossier_candidat, exist_ok=True)

    data_dict[candidat] = {}

    for fichier in fichiers_candidat:

        chemin = os.path.join(dossier, fichier)

        # Charger données
        df = pd.read_csv(chemin, header=None, skiprows=1)

        # Traitement EDA
        signals, info = nk.eda_process(df[0], sampling_rate=4)

        # Identifier bloc
        if "bloc1" in fichier:
            bloc = "bloc1"
        elif "bloc2" in fichier:
            bloc = "bloc2"
        elif "bloc3" in fichier:
            bloc = "bloc3"
        else:
            continue

        # ======================
        # STOCKAGE EN MÉMOIRE
        # ======================
        data_dict[candidat][bloc] = signals

        # ======================
        # SAUVEGARDE CSV
        # ======================
        chemin_csv = os.path.join(dossier_candidat, f"{bloc}_signals.csv")
        signals.to_csv(chemin_csv, index=False)

        # ======================
        # PLOT
        # ======================
        nk.eda_plot(signals, info)

        plt.ylim(signals["EDA_Clean"].min() - 0.05,
                 signals["EDA_Clean"].max() + 0.05)

        plt.suptitle(f"EDA - {candidat}_{bloc}_V2",
                     fontsize=14, fontweight='bold')

        chemin_plot = os.path.join(dossier_candidat, f"{bloc}.png")
        plt.savefig(chemin_plot, dpi=300, bbox_inches='tight')

        plt.close()

    nb_candidats_ok += 1

# ============================
# RÉSUMÉ
# ============================
print("\n====================")
print(f"Candidats conservés : {nb_candidats_ok}")
print(f"Dictionnaire créé avec {len(data_dict)} candidats")
print("====================")
```
